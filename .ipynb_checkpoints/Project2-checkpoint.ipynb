```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('crime-housing-austin-2015.csv')
df.columns
```

```python
# 1. Clean the columns by removing strings/symbols and converting to numeric
df['Medianhomevalue'] = df['Medianhomevalue'].astype(str).replace(r'[\$,]', '', regex=True)
df['Medianhomevalue'] = pd.to_numeric(df['Medianhomevalue'], errors='coerce')

df['Percentageofhomeswithin1/4-mioftransitstop'] = df['Percentageofhomeswithin1/4-mioftransitstop'].astype(str).replace(r'[%]', '', regex=True)
df['Percentageofhomeswithin1/4-mioftransitstop'] = pd.to_numeric(df['Percentageofhomeswithin1/4-mioftransitstop'], errors='coerce')

# Drop missing values in these two columns to ensure clean math
plot_df = df.dropna(subset=['Percentageofhomeswithin1/4-mioftransitstop', 'Medianhomevalue'])

# 2. Calculate Averages and Standard Deviations
home_avg = plot_df['Medianhomevalue'].mean()
home_std = plot_df['Medianhomevalue'].std()
transit_avg = plot_df['Percentageofhomeswithin1/4-mioftransitstop'].mean()
transit_std = plot_df['Percentageofhomeswithin1/4-mioftransitstop'].std()

print("--- Descriptive Statistics ---")
print(f"Median Home Value:      Average = ${home_avg:,.2f} | Std Dev = ${home_std:,.2f}")
print(f"Transit Accessibility:  Average = {transit_avg:.2f}%  | Std Dev = {transit_std:.2f}%\n")

# 3. Calculate Pearson Correlation (Coefficient only via Pandas)
pearson_corr = plot_df['Percentageofhomeswithin1/4-mioftransitstop'].corr(plot_df['Medianhomevalue'], method='pearson')
print(f"--- Correlation ---")
print(f"Pearson Correlation (r): {pearson_corr:.4f}")
print("(Note: To calculate the p-value or run a t-test, you will need to import scipy.stats)\n")

# 4. Create the Scatterplot
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=plot_df, 
    x='Percentageofhomeswithin1/4-mioftransitstop', 
    y='Medianhomevalue',
    alpha=0.5,
    color='#2b7bba',
    edgecolor=None
)

# Formatting the plot
plt.title('Housing Market vs. Transit Accessibility in Austin', fontsize=14, pad=15)
plt.xlabel('Homes within 1/4-mile of a Transit Stop (%)', fontsize=12)
plt.ylabel('Median Home Value ($)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()

plt.show()
```
